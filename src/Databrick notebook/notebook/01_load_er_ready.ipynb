```python
catalog = "workspace"
schema = "entity_resolution_project"
volume = "er_project_files"

project_prefix = f"{catalog}.{schema}"
base_path = f"/Volumes/{catalog}/{schema}/{volume}"

er_ready_path = f"{base_path}/data_er_ready.csv"
er_ready_table = f"{project_prefix}.company_er_ready"


df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("multiLine", True)
    .option("quote", '"')
    .option("escape", '"')
    .csv(er_ready_path)
)

display(df)

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(er_ready_table)
)
```
