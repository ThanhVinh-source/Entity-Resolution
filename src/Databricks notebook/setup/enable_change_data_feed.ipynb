# Enable Change Data Feed for GraphRAG Documents
This setup notebook enables Delta Change Data Feed on the GraphRAG document table, configures retention for deleted files, and verifies the resulting table properties.

```python
# Enable Delta Change Data Feed on the GraphRAG documents table.
spark.sql("""
ALTER TABLE workspace.entity_resolution_project.company_er_graphrag_documents
SET TBLPROPERTIES (delta.enableChangeDataFeed = true)
""")
```

```python
# Set deleted-file retention for Change Data Feed history.
spark.sql("""
ALTER TABLE workspace.entity_resolution_project.company_er_graphrag_documents
SET TBLPROPERTIES (
  'delta.deletedFileRetentionDuration' = 'interval 30 days'
)
""")
```

```python
# Verify the GraphRAG documents table properties.
display(spark.sql("""
SHOW TBLPROPERTIES workspace.entity_resolution_project.company_er_graphrag_documents
"""))
```
